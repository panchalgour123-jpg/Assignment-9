# Question 1: NoSQL vs Relational Databases

#   What are the key differences between relation databases and MongoDB?

**Answer:-**
  **NoSQL vs Relational Databases**

 **a) Key differences between Relational Databases and MongoDB**

Relational databases like MySQL and Oracle store data in tables with rows and columns. They have a fixed structure (schema) and use SQL language to manage data.

MongoDB is a NoSQL database that stores data in documents (JSON-like format). It does not require a fixed structure, so data can be stored in a more flexible way.

**Main differences:**

* Relational databases use tables, while MongoDB uses collections and documents.

* SQL databases have a fixed schema, while MongoDB has a flexible schema.

* SQL databases are good for complex relationships and transactions.

* MongoDB is good for handling large and changing data.

# b) When is MongoDB preferred over SQL systems?

:-**MongoDB is preferred when:**

* The data structure changes frequently.
We need to store large amounts of data.
Fast development is required.

* The application needs high scalability.
* Data is unstructured or semi-structured.
* Examples include social media applications, real-time applications, and content management systems.

# c) Explain schema flexibility and its risks.

:-Schema flexibility means MongoDB allows storing different types of documents without following a fixed structure. Developers can easily add or remove fields whenever required.

**Advantages:**

* Easy to update data structure.
* Faster application development.
* Suitable for changing requirements.

**Risks:**

* Data may become inconsistent.
* Data validation becomes difficult.
* Poor design can affect database performance.

# d) How does embedding vs referencing impact performance?

:- Embedding means storing related data inside one document. It provides faster access because all data can be retrieved in one query. It is useful when data is small and frequently used together.

Referencing means storing related data separately and connecting it using references. It reduces data duplication but may require multiple queries.

Embedding improves read performance, while referencing is better for large and frequently changing data.



# Qestion 2: Indexing, Sharding & Replication

# a) How do indexes improve MongoDB query performance?

**Answer:-**

:- Indexes help MongoDB find data faster without checking every document in a collection. They store data in a sorted way, which makes searching, filtering, and sorting operations quicker. Without indexes, MongoDB has to scan all documents, which takes more time.

# b) What is sharding? When is it required?

:- Sharding is a method of dividing a large database into smaller parts called shards and storing them on different servers. It is used when the amount of data becomes very large or when a single server cannot handle the storage and workload. Sharding improves performance and allows the database to handle more users and data.

# c) What is replication and how does it ensure high availability?

:- Replication means keeping multiple copies of the same data on different servers. In MongoDB, this is done using a Replica Set. If the main server fails, another server with a copy of the data can become the primary server. This keeps the database available and prevents data loss.

# d) How does improper shard key selection impact performance?

:- A shard key decides how data is distributed among different shards. If the shard key is chosen incorrectly, data may not be distributed equally, causing some servers to become overloaded while others remain less used. This can reduce performance and slow down database operations.

# Question 4: Random Variables & Bayes

# a) What is a random variable?

:-A random variable is a variable whose value depends on the result of a random experiment. It assigns numerical values to possible outcomes of an event. For example, the number of customers visiting a shop in a day can be considered a random variable because the number may change every day.

#b) Difference between discrete and continuous random variables

:-A discrete random variable takes specific, countable values. It usually represents things that can be counted, such as the number of products sold, number of employees, or number of customers.
A continuous random variable can take any value within a range. It is usually measured rather than counted, such as customer waiting time, product weight, or temperature.

#c) Explain covariance and its interpretation

:-Covariance shows the relationship between two variables and indicates whether they move together or in opposite directions.
If covariance is positive, both variables increase or decrease together. For example, higher advertising spending may lead to higher sales.
If covariance is negative, one variable increases while the other decreases. For example, higher product prices may reduce demand.
If covariance is close to zero, it means there is little or no relationship between the variables.

```python
db.orders.createIndex({ customer_id: 1, order_date: 1 })
```

**3. How to check if the index is being used:**

MongoDB provides the explain() method to check query performance and index usage.

**Example Query:**


```python
db.orders.find({customer_id: 101, order_date: "2026-07-01"}).explain("executionStats")
```

In the output, if MongoDB uses the created index, it will show IXSCAN (Index Scan). If it shows COLLSCAN (Collection Scan), MongoDB is scanning the whole collection and the index is not being used.

# Question 10: Hypothesis Testing of Average Monthly Spending

# 1. How would you statistically test this claim?
Answer:-

To test the company’s claim that the average monthly spending is ₹5000, we can use hypothesis testing. First, collect a sample of customers’ monthly spending data and calculate the sample mean. Then, compare this sample mean with ₹5000 using a suitable statistical test such as a one-sample t-test (when the population standard deviation is unknown). The test result will show whether there is enough evidence to accept or reject the company’s claim.

#2. Define null and alternative hypothesis

:- The null hypothesis (H₀) represents the company’s claim that the average monthly spending is ₹5000.

H₀: μ = ₹5000

The alternative hypothesis (H₁) represents that the average monthly spending is different from ₹5000.

H₁: μ ≠ ₹5000

If the test results provide strong evidence against the null hypothesis, we reject H₀ and conclude that the average spending is not ₹5000.

# 3. What type of error is more dangerous in business — Type I or Type II?

:- In business, the importance of Type I and Type II errors depends on the situation. Generally, Type II error can be more dangerous because it means failing to identify a real problem or opportunity. For example, if a company’s average customer spending has actually decreased but the test fails to detect it, the company may continue making wrong business decisions.

However, in some cases, Type I error can also be costly, such as launching a product or making a major decision based on incorrect evidence. Therefore, businesses choose the error level based on the potential impact of wrong decisions.
